# Stochastic Projected Gross-Pitaevskii (SPGP) simulation
This example uses the module sgp (stochastic gross-piteavskii) to simulate the formation of a two-component BEC at finite temperature. To simulate condensation we artificially ramp the chemical potential from zero to above its critical value and let the system evolve until equilibrium. We employ the following SPGP:
$$
i\hbar\frac{\partial}{\partial t} \psi_j(\mathbf{r}, t) = \mathcal{P}\left\{ (1-i\gamma)\left[\left(-\frac{\hbar^2\nabla^2}{2m} + g|\psi_j(\mathbf{r}, t)|^2 + g_s|\psi_{3-j}|^2 - \mu(t)\right)\psi_j\right] + \eta_j(\mathbf{r}, t)\right\}
$$

where:
 - $\psi_j(\mathbf{r}, t)$ is the wave-function of the $j$-component.
 - $\mathcal{P}$ is a low energy projector that removes energies above a cutoff $\Lambda$
 - $\gamma$ is the discipative coefficient
 - $\eta_j(\mathbf{r}, t)$

```python
import sgp
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams.update({
    'text.usetex': True,
    'font.family': 'serif',
    'font.serif': ['Computer Modern'],
    'axes.labelsize': 12,   # Font size for axis labels
    'font.size': 12,        # General font size
    'legend.fontsize': 12,  # Font size for legend
    'xtick.labelsize': 12,  # Font size for x-axis tick labels
    'ytick.labelsize': 12   # Font size for y-axis tick labels
})
plt.rcParams["figure.figsize"] = (4, 3)
plt.rc('text', usetex=True)
plt.rcParams['text.latex.preamble'] = r'\usepackage{amsmath, amssymb, bm}'
```

```python
nx = 2**7
ny = 2**7

lx = 30
ly = 30

dt = 2e-4

# Chemical potential schedule
def mu_fun(t, tau, mu_i=0.1, mu_f=20.0, t_r=10):
    return (mu_i * (t <= t_r)
            + (mu_i + (mu_f - mu_i) * (t - t_r) / tau) * (t > t_r) * (t <= (t_r + tau))
            + mu_f * (t > (t_r + tau)))

geo = sgp.Geometry(N=(nx, ny), L=(lx, ly))
dispersion = (geo.kx_**2 + geo.ky_**2) / 2

syst = sgp.System(g_2d=1.0, spin_g_factor=0.5, gamma=0.03, temp=0.2)
syst.set_geometry(geo)
syst.set_dispersion(dispersion)

psi1_initial = 0e-2 * (np.random.default_rng(1).normal(size=(nx, ny)) + 1j * np.random.default_rng(1).normal(size=(nx, ny)))
psi2_initial = 0e-2 * (np.random.default_rng(2).normal(size=(nx, ny)) + 1j * np.random.default_rng(2).normal(size=(nx, ny)))
```
